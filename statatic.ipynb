```python
import pandas as pd
```

```python
df = pd.read_csv("categories.csv")
df.head()
```

```python
leaf_category = df[df["children_count"] == 0]
print(len(leaf_category) , " / " , len(df))
print(leaf_category.head())
```

```python
full_url_link = leaf_category[["id", "name", "product_count"]].copy()

full_url_link["link"] = (
    "https://tiki.vn/"
    + leaf_category["url_key"].astype(str)
    + "/c"
    + leaf_category["id"].astype(str)
)

full_url_link.head()
```

```python
full_url_link = full_url_link.sort_values(["product_count", "id"], ascending=[True, True])

full_url_link.to_csv(
    "full_url_link.csv",
    index=False,
    encoding="utf-8-sig"
)
```
